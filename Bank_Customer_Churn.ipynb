```python
df.isnull().sum()
```

```python
# filling India in all country's null value.
df['country'] = df['country'].fillna('India')
```

```python
# total number of female records in the dataset
df['gender'].value_counts()
```

```python
# Compare the count of male and female, and fill all missing values in the Gender column with the gender that has the higher count.
df['gender'] = df['gender'].fillna('Male')
```

```python
# Analyze the Age column and fill all missing age values using average age. Ensure that the age values are integer.
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['age'] = df['age'].astype('int')
df['age'].dtype
```

```python
# Fill the first 500 missing values with the average balance
df['balance'] = df['balance'].fillna(df['balance'].mean(),limit=500)
```

```python
# Fill the next 500 missing values with the maximum balance
df['balance'] = df['balance'].fillna(df['balance'].max(),limit=500)
```

```python
# Fill the remaining missing values with 0
df['balance'] = df['balance'].fillna(0)
```

```python
import numpy as np
```

```python
# Ensure the balance values contain exactly 2 decimal places
df['balance'] = np.round(df['balance'],2)
```

```python
# For the Active_Member column, fill missing values such that 50% of them are assigned 1 (active)
1028/2 # 50% of missing values
df['active_member'] = df['active_member'].fillna(1,limit=514)
```

```python
# the remaining 50% are assigned 0 (inactive)
df['active_member'] = df['active_member'].fillna(0)
```

```python
# the average value of the Estimated_Salary column and fill all missing salary values with this average 
df['estimated_salary'] = df['estimated_salary'].fillna(df['estimated_salary'].mean())
```

```python
# ensuring the values have exactly 2 decimal places
df['estimated_salary'] = np.round(df['estimated_salary'],2)
```
